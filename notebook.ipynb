```python
# extracting presidential election records
candidate_columns = [
    "ANO_ELEICAO",
    "NR_TURNO",
    "CD_ELEICAO",
    "DT_ELEICAO",
    "SG_UF",
    "CD_MUNICIPIO",
    "NM_MUNICIPIO",
    "NR_ZONA",
    "CD_CARGO",
    "DS_CARGO",
    "SQ_CANDIDATO",
    "NR_CANDIDATO",
    "NM_CANDIDATO",
    "NM_URNA_CANDIDATO",
    "SG_PARTIDO",
    "QT_VOTOS_NOMINAIS",
    "QT_VOTOS_NOMINAIS_VALIDOS"
]

candidate_chunks = []

for chunk in pd.read_csv(
    candidate_file,
    sep=";",
    encoding="latin1",
    chunksize=100000,
    usecols=candidate_columns
):
    filtered_chunk = chunk[
        (chunk["DS_CARGO"] == "Presidente") &
        (chunk["NR_TURNO"].isin([1, 2]))
    ]

    candidate_chunks.append(filtered_chunk)

df_candidatepres = pd.concat(candidate_chunks, ignore_index=True)
```

```python
# checking candidates and duplicate records
print("Candidates by round:")
display(
    df_candidatepres.groupby("NR_TURNO")[
        ["NR_CANDIDATO", "NM_URNA_CANDIDATO", "SG_PARTIDO"]
    ].apply(lambda x: x.drop_duplicates().sort_values("NR_CANDIDATO"))
)

duplicates = df_candidatepres.duplicated(
    subset=[
        "SG_UF",
        "CD_MUNICIPIO",
        "NR_ZONA",
        "NR_TURNO",
        "SQ_CANDIDATO"
    ]
)

print("\nDuplicated records:", duplicates.sum())
```

```python
# validating candidate votes against the details dataset

keys = ["SG_UF", "CD_MUNICIPIO", "NR_ZONA", "NR_TURNO"]

candidate_votes_zone = (
    df_candidatepres
    .groupby(keys, as_index=False)["QT_VOTOS_NOMINAIS_VALIDOS"]
    .sum()
)

detail_votes_zone = df_detailpres[
    keys + ["QT_VOTOS_NOMINAIS_VALIDOS"]
].copy()

vote_check = candidate_votes_zone.merge(
    detail_votes_zone,
    on=keys,
    how="outer",
    suffixes=("_candidate", "_detail"),
    indicator=True
)

print("Records in both datasets:", (vote_check["_merge"] == "both").sum())
print("Records only in candidate dataset:", (vote_check["_merge"] == "left_only").sum())
print("Records only in details dataset:", (vote_check["_merge"] == "right_only").sum())

matched = vote_check[vote_check["_merge"] == "both"]

differences = matched[
    matched["QT_VOTOS_NOMINAIS_VALIDOS_candidate"] !=
    matched["QT_VOTOS_NOMINAIS_VALIDOS_detail"]
]

print("Different vote totals:", len(differences))
```

```python
# aggregating candidate votes by municipality and round

candidate_group_columns = [
    "NR_TURNO",
    "SG_UF",
    "CD_MUNICIPIO",
    "NM_MUNICIPIO",
    "NR_CANDIDATO",
    "NM_URNA_CANDIDATO",
    "SG_PARTIDO"
]

df_candidatepres_mun = (
    df_candidatepres
    .groupby(candidate_group_columns, as_index=False)[
        ["QT_VOTOS_NOMINAIS", "QT_VOTOS_NOMINAIS_VALIDOS"]
    ]
    .sum()
)

print(df_candidatepres_mun.shape)
df_candidatepres_mun.head()
```

```python
# merging total valid votes and calculating vote share

df_candidatepres_mun = df_candidatepres_mun.merge(
    df_detailpres_mun[
        ["SG_UF", "CD_MUNICIPIO", "NR_TURNO", "QT_TOTAL_VOTOS_VALIDOS"]
    ],
    on=["SG_UF", "CD_MUNICIPIO", "NR_TURNO"],
    how="left"
)

df_candidatepres_mun["VOTE_SHARE"] = (
    df_candidatepres_mun["QT_VOTOS_NOMINAIS_VALIDOS"] /
    df_candidatepres_mun["QT_TOTAL_VOTOS_VALIDOS"] * 100
)

df_candidatepres_mun.head()
```

```python
# checking municipality coverage and missing values

print("Records by round:")
print(df_candidatepres_mun["NR_TURNO"].value_counts().sort_index())

municipality_rounds = (
    df_candidatepres_mun
    .groupby(["SG_UF", "CD_MUNICIPIO"])["NR_TURNO"]
    .nunique()
)

print("\nMunicipalities with both rounds:", (municipality_rounds == 2).sum())
print("Municipalities with only one round:", (municipality_rounds == 1).sum())

print("\nMissing vote shares:", df_candidatepres_mun["VOTE_SHARE"].isna().sum())
```

```python
# selecting runoff candidates and valid vote shares

df_runoff = df_candidatepres_mun[
    df_candidatepres_mun["NR_CANDIDATO"].isin([13, 22])
].copy()

print(df_runoff.shape)
print(df_runoff.groupby(["NR_TURNO", "NR_CANDIDATO"]).size())
```

```python
# reshaping candidate results for round comparison

df_vote_comparison = df_runoff.pivot(
    index=["SG_UF", "CD_MUNICIPIO", "NM_MUNICIPIO"],
    columns=["NR_TURNO", "NR_CANDIDATO"],
    values="VOTE_SHARE"
)

df_vote_comparison.columns = [
    f"ROUND_{turn}_CANDIDATE_{candidate}"
    for turn, candidate in df_vote_comparison.columns
]

df_vote_comparison = df_vote_comparison.reset_index()

print(df_vote_comparison.shape)
df_vote_comparison.head()
```

```python
# calculating percentage point changes

df_vote_comparison["LULA_CHANGE_PP"] = (
    df_vote_comparison["ROUND_2_CANDIDATE_13"] -
    df_vote_comparison["ROUND_1_CANDIDATE_13"]
)

df_vote_comparison["FLAVIO_CHANGE_PP"] = (
    df_vote_comparison["ROUND_2_CANDIDATE_22"] -
    df_vote_comparison["ROUND_1_CANDIDATE_22"]
)

df_vote_comparison.head()
```

```python
# ranking municipalities by candidate vote share changes

print("Top 10 municipalities: Lula gains")
display(
    df_vote_comparison
    .sort_values("LULA_CHANGE_PP", ascending=False)
    [["SG_UF", "NM_MUNICIPIO", "LULA_CHANGE_PP"]]
    .head(10)
)

print("Top 10 municipalities: Lula losses")
display(
    df_vote_comparison
    .sort_values("LULA_CHANGE_PP")
    [["SG_UF", "NM_MUNICIPIO", "LULA_CHANGE_PP"]]
    .head(10)
)
```

```python
# aggregating overseas votes by country, round and candidate group

df_country_rounds = (
    df_abroad_all
    .groupby(["COUNTRY", "NR_TURNO", "CANDIDATE_GROUP"])
    ["QT_VOTOS_NOMINAIS_VALIDOS"]
    .sum()
    .unstack(["NR_TURNO", "CANDIDATE_GROUP"], fill_value=0)
)

# creating consistent columns for both rounds
expected_columns = pd.MultiIndex.from_product(
    [[1, 2], ["Lula", "Flávio Bolsonaro", "Others"]]
)

df_country_rounds = df_country_rounds.reindex(
    columns=expected_columns,
    fill_value=0
)

df_country_rounds.columns = [
    f"R{round_number}_{group.upper().replace(' ', '_')}"
    for round_number, group in df_country_rounds.columns
]

df_country_rounds = df_country_rounds.reset_index()

print(df_country_rounds.head())
print("Total countries:", df_country_rounds["COUNTRY"].nunique())
```

```python
# calculating totals, percentages and winners for both rounds

for round_number in [1, 2]:

    lula = f"R{round_number}_LULA"
    flavio = f"R{round_number}_FLÁVIO_BOLSONARO"
    others = f"R{round_number}_OTHERS"

    total = f"R{round_number}_TOTAL_VOTES"
    winner = f"R{round_number}_WINNER"

    df_country_rounds[total] = (
        df_country_rounds[lula]
        + df_country_rounds[flavio]
        + df_country_rounds[others]
    )

    df_country_rounds[winner] = np.select(
        [
            df_country_rounds[total] == 0,
            (df_country_rounds[lula] > df_country_rounds[flavio]) &
            (df_country_rounds[lula] > df_country_rounds[others]),
            (df_country_rounds[flavio] > df_country_rounds[lula]) &
            (df_country_rounds[flavio] > df_country_rounds[others])
        ],
        ["No valid votes", "Lula", "Flávio Bolsonaro"],
        default="Others"
    )

    for candidate, votes in [
        ("LULA", lula),
        ("FLÁVIO_BOLSONARO", flavio),
        ("OTHERS", others)
    ]:
        df_country_rounds[f"R{round_number}_{candidate}_PCT"] = (
            df_country_rounds[votes]
            .div(df_country_rounds[total].replace(0, np.nan))
            .mul(100)
            .fillna(0)
            .round(2)
        )

print(df_country_rounds.head(10))
```
